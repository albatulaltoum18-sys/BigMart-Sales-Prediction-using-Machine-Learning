```python
outlet_types = ['Grocery Store', 'Supermarket Type1', 'Supermarket Type2', 'Supermarket Type3']
low_fat_sales = [705, 3609, 598, 605]
regular_fat_sales = [378, 1968, 330, 330]

# Position of bars on the x-axis
x = np.arange(len(outlet_types))

# Width of the bars
width = 0.35

# Plotting the bars
fig, ax = plt.subplots(figsize=(10,6))
bars1 = ax.bar(x - width/2, low_fat_sales, width, label='Low Fat', color='#1f77b4')
bars2 = ax.bar(x + width/2, regular_fat_sales, width, label='Regular Fat', color='#ff7f0e')

# Adding labels and title
ax.set_xlabel('Outlet Type', fontsize=14)
ax.set_ylabel('Number of Sales', fontsize=14)
ax.set_title('Sales Comparison of Low Fat vs Regular Fat by Outlet Type', fontsize=16)
ax.set_xticks(x)
ax.set_xticklabels(outlet_types, fontsize=12)
ax.legend()

# Adding value labels on top of bars
def add_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

add_labels(bars1)
add_labels(bars2)

# Display plot
plt.tight_layout()
plt.show()
```

```python
def show_nulls_information(data, threshold = 0.8):
    nulls_count_df = data.isnull().sum().reset_index().rename(columns = {0: 'Count'})
    nulls_count_df = nulls_count_df[nulls_count_df['Count'] > 0].sort_values(by = 'Count',ascending = False)
    nulls_count_df['Percentage'] = nulls_count_df['Count'] / data.shape[0]
    display(nulls_count_df)
    print('-' * 50)
    # Lets Get Features To Delete
    features_to_drop = nulls_count_df[nulls_count_df['Percentage'] >= threshold]['index'].values
    print(f'Features To Drop : {features_to_drop}')
    # Lets Get Features To Fill
    features_to_fill = nulls_count_df[nulls_count_df['Percentage'] < threshold]['index'].values
    print('-' * 50)
    print(f'Features To Fill : {features_to_fill}')
    return features_to_drop, features_to_fill

features_to_drop, features_to_fill = show_nulls_information(data)
```

</div>

```python
def get_redundant_features(data, threshold = 0.8):
    redundant_features = []
    for col in data.columns:
        count_value = (data.groupby(col).count().iloc[:, 0].max()) / data.shape[0]
        if count_value > threshold:
            print(f'{col}Is Redundant Features')
            redundant_features.append(col)
    return redundant_features

redundant_features = get_redundant_features(data)
print(f'Number Of Redundant Features : {len(redundant_features)}')
```

<div style="background-color: #22b2da; text-align: center; padding: 10px 20px; color: #fff; border-radius: 25px; display: inline-block;">
    <h2>Handle Correlation Between Features</h2>
</div>

```python
def get_correlation_features(data, target, threshold = 0.7):
    num_features = list(data.select_dtypes(exclude = ['O', 'category']).columns)
    num_features.remove(target)
    corr_matrix = data[num_features].corr()
    features_correlated = []
    for i in range(corr_matrix.shape[0]):
        for j in range(i+1, corr_matrix.shape[0]):
            corr_coeff = np.abs(corr_matrix.iloc[i, j])
            if corr_coeff >= threshold:
                corr_coeff_i = np.abs(np.corrcoef(data[corr_matrix.index[i]], data[target])[0, 1])
                corr_coeff_j = np.abs(np.corrcoef(data[corr_matrix.columns[j]], data[target])[0, 1])
                if corr_coeff_i >= corr_coeff_j:
                    print(f'We Will Drop {corr_matrix.columns[j]}')
                    features_correlated.append(corr_matrix.columns[j])
                else :
                    print(f'We Will Drop {corr_matrix.index[i]}')
                    features_correlated.append(corr_matrix.index[i])
    return features_correlated
```

    <h2>Handle Outliers</h2>
</div>

```python
def clamp_outleris(data):
    num_features = list(data.select_dtypes(exclude = ['O', 'category']).columns)
    for col in num_features:
        Q1, Q3 = np.quantile(data[col], [0.25, 0.75])
        IQR = Q3 - Q1
        upper_bound = Q3 + (1.5 * IQR)
        lower_bound = Q1 - (1.5 * IQR)

        idx_upper_bound = data[data[col] > upper_bound].index
        data.loc[idx_upper_bound, col] = upper_bound

        idx_lower_bound = data[data[col] < lower_bound].index
        data.loc[idx_lower_bound, col] = lower_bound
    return data
```

    <h2>Handle Skewness Features</h2>
</div>

```python
def skewness_coeff(data):
    num_features = list(data.select_dtypes(exclude = ['O', 'category']).columns)
    sk_coeff_dict = {}
    for col in num_features:
        sk_coeff = np.sum((data[col] - np.mean(data[col])) ** 3) / ((data.shape[0] - 1) * (np.std(data[col]) ** 3))
        print(f'Skewness Coefficient Of {col} : {sk_coeff}')
        sk_coeff_dict[col] = sk_coeff
    return sk_coeff_dict

sk_coeff_dict = skewness_coeff(train_df)
```

    <h2>Discovery of the features follows the normal distribution</h2>

```python
from scipy.stats import shapiro
def check_normall_dist(data, alpha = 1.76e-20):
    num_features = list(data.select_dtypes(exclude = ['O', 'category']).columns)
    features_follow_normall = []
    features_not_follow_normall = []
    for col in num_features:
        w, p = shapiro(data[col])
        print(f'Shapiro Wilk Test On {col}')
        print(f'W = {w}, P-value = {p}')
        if p > alpha:
            print(f'{col} looks like Normall Distribution (Fail To Reject H0)')
            features_follow_normall.append(col)
        else :
            print(f'{col} Not looks like Normall Distribution (Reject H0)')
            features_not_follow_normall.append(col)
        print('-' * 100)
    return features_follow_normall, features_not_follow_normall
```

    <h2>Build Model</h2>
</div>

```python
def evaluate_model(true, predicted):
    mae = mean_absolute_error(true, predicted)
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(true, predicted)
    return mae, mse, rmse, r2
```

```python
models = {
        'Linear Regression': LinearRegression(),
        'Lasso': Lasso(),
        'Ridge': Ridge(),
        'ElasticNet': ElasticNet(),
        'KNN': KNeighborsRegressor(),
        'SVR': SVR(),
        'Decision Tree': DecisionTreeRegressor(),
        'Random Forest': RandomForestRegressor(),
        'AdaBoostRegressor': AdaBoostRegressor(),
        'GradientBoostingRegressor': GradientBoostingRegressor(),
        'XGBRegressor': XGBRegressor()
}

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)

    model_train_mae, model_train_mse, model_train_rmse, model_train_r2 = evaluate_model(y_train, y_train_pred)
    model_test_mae, model_test_mse, model_test_rmse, model_test_r2 = evaluate_model(y_val, y_val_pred)

    print(list(models.keys())[i])

    print()
    print('Model Performance In Training Set')
    print(f'- MAE : {model_train_mae}')
    print(f'- MSE : {model_train_mse}')
    print(f'- RMSE : {model_train_rmse}')
    print(f'- R2 : {model_train_r2}')
    print('-'* 40)
    print('Model Performance In Test Set')
    print(f'- MAE : {model_test_mae}')
    print(f'- MSE : {model_test_mse}')
    print(f'- RMSE : {model_test_rmse}')
    print(f'- R2 : {model_test_r2}')
    print('='* 40)
    print()
```

```python
from sklearn.model_selection import RandomizedSearchCV

ridge_params = {
    'alpha': [0.01, 0.1, 1.0, 10.0, 100.0],
    'solver': ['auto', 'svd', 'cholesky', 'lsqr']
}

svr_params = {
    'C': [0.1, 1.0, 10.0, 100.0],
    'kernel': ['linear', 'poly', 'rbf', 'sigmoid'],
    'gamma': ['scale', 'auto', 0.01, 0.1]
}

decision_tree_params = {
    'max_depth': [None, 3, 5, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [1.0, 'sqrt', 'log2']
}

random_forest_params = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [1.0, 'sqrt', 'log2']
}

gradient_boosting_params = {
    'learning_rate': [0.01, 0.1, 0.2],
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 5, 7],
    'min_samples_split': [2, 5, 10],
    'subsample': [0.5, 0.75, 1.0]
}

xgb_regressor_params = {
    'learning_rate': [0.01, 0.1, 0.2],
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 5, 7],
    'subsample': [0.5, 0.75, 1.0]
}

randomcv_models = [
    ('Ridge', Ridge(), ridge_params),
    ('SVR', SVR(), svr_params),
    ('Decision Tree', DecisionTreeRegressor(), decision_tree_params),
    ('Random Forest', RandomForestRegressor(), random_forest_params),
    ('Gradient Boosting', GradientBoostingRegressor(), gradient_boosting_params),
    ('XGBoost', XGBRegressor(), xgb_regressor_params)
]

model_param = {}
for name, model, params in randomcv_models:
    random = RandomizedSearchCV(estimator=model, 
                               param_distributions=params,
                               n_iter=100, 
                               cv = 3,
                               verbose=1,
                               n_jobs=-1)
    random.fit(X_train, y_train.ravel())
    model_param[name] = random.best_params_

for model_name in model_param:
    print(f'------------------ Best Params For {model_name} ------------------')
    print(model_param[model_name])
```

```python
models = {
        'Ridge': Ridge(solver = 'lsqr', alpha = 1.0),
        'SVR': SVR(kernel='linear', gamma='scale', C = 100.0),
        'Decision Tree': DecisionTreeRegressor(min_samples_split = 2, min_samples_leaf = 2, max_features = 1.0, max_depth = 5),
        'Random Forest': RandomForestRegressor(n_estimators = 100, min_samples_split = 5, min_samples_leaf = 4, max_features = 1.0, max_depth = 10),
        'GradientBoostingRegressor': GradientBoostingRegressor(subsample = 1.0, n_estimators = 50, min_samples_split = 10, max_depth = 3, learning_rate = 0.1),
        'XGBRegressor': XGBRegressor(subsample = 1.0, n_estimators = 50, max_depth = 3, learning_rate = 0.1)
}

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)

    model_train_mae, model_train_mse, model_train_rmse, model_train_r2 = evaluate_model(y_train, y_train_pred)
    model_test_mae, model_test_mse, model_test_rmse, model_test_r2 = evaluate_model(y_val, y_val_pred)

    print(list(models.keys())[i])

    print()
    print('Model Performance In Training Set')
    print(f'- MAE : {model_train_mae}')
    print(f'- MSE : {model_train_mse}')
    print(f'- RMSE : {model_train_rmse}')
    print(f'- R2 : {model_train_r2}')
    print('-'* 40)
    print('Model Performance In Test Set')
    print(f'- MAE : {model_test_mae}')
    print(f'- MSE : {model_test_mse}')
    print(f'- RMSE : {model_test_rmse}')
    print(f'- R2 : {model_test_r2}')
    print('='* 40)
    print()
```
